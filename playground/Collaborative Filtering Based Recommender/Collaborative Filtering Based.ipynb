## **Importing Libraries**

```python
import numpy as np
import pandas as pd
```

## **Loading Datasets**

```python
books = pd.read_csv('Datasets/books.csv')
users = pd.read_csv('Datasets/users.csv')
ratings = pd.read_csv('Datasets/ratings.csv')
```

```python
## Joining the Books dataset with ratings on "ISBN"

ratings_with_name = ratings.merge(books,on='ISBN')
```

## **Collaborative Filtering Based Recommender System**


We are only considering the users who has provided more than 200 reviews

```python
x = ratings_with_name.groupby('User-ID')['Book-Rating'].count() > 200
useres_with_more_than_200_rating = x[x].index
```

```python
filtered_rating = ratings_with_name[ratings_with_name['User-ID'].isin(useres_with_more_than_200_rating )]
```

```python
filtered_rating.head()
```

Now, We Will Consider the books that were rated at least 50 times !

```python
y = filtered_rating.groupby('Book-Title').count()['Book-Rating']>=50
famous_books = y[y].index
```

```python
final_ratings = filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
final_ratings.head()
```

```python
pt = final_ratings.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
```

```python
pt
```

```python
pt.fillna(0,inplace=True)
```

```python
pt
```

## **Cosine Similarity**

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_scores = cosine_similarity(pt)
```

```python
similarity_scores.shape
```

```python
def recommend(book_name):
    # index fetch
    index = np.where(pt.index==book_name)[0][0]
    similar_items = sorted(list(enumerate(similarity_scores[index])),key=lambda x:x[1],reverse=True)[1:6]
    
    data = []
    for i in similar_items:
        item = []
        temp_df = books[books['Book-Title'] == pt.index[i[0]]]
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Title'].values))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Author'].values))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Image-URL-M'].values))
        
        data.append(item)
    
    return data
```

```python
recommend('Harry Potter and the Prisoner of Azkaban (Book 3)')
```

## **KNN**

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np

model = NearestNeighbors(n_neighbors=10, metric='euclidean')
model.fit(pt)
distances, indices = model.kneighbors(pt)
similarity_scores = 1 / (1 + distances / np.max(distances))  # Normalize similarities to [0, 1]
```

```python
from IPython.display import HTML
import numpy as np

def recommend(book_name):
    index = np.where(pt.index == book_name)[0][0]
    similar_indices = indices[index][1:5]  # Get top 4 similar books (excluding self)
    similar_scores = similarity_scores[index][1:5]
    
    selected_book = books[books['Book-Title'] == book_name].drop_duplicates('Book-Title')
    if selected_book.empty:
        print(f"Book '{book_name}' not found in books dataset.")
        return
    selected_title = selected_book['Book-Title'].values[0]
    selected_author = selected_book['Book-Author'].values[0]
    selected_image = selected_book['Image-URL-M'].values[0]
    
    html = f"""
    <h2>📚 Selected Book</h2>
    <div style='text-align: center; margin-bottom: 40px;'>
        <img src="{selected_image}" style="height:250px;"><br>
        <b>{selected_title}</b><br>
        <i>{selected_author}</i>
    </div>
    <h2>📚 Recommended Books</h2>
    <div style='display: flex; flex-wrap: wrap;'>
    """
    
    for i in similar_indices:
        temp_df = books[books['Book-Title'] == pt.index[i]].drop_duplicates('Book-Title')
        if temp_df.empty:
            continue
        title = temp_df['Book-Title'].values[0]
        author = temp_df['Book-Author'].values[0]
        image = temp_df['Image-URL-M'].values[0]
        
        html += f"""
        <div style="width: 22%; margin: 1%; text-align: center;">
            <img src="{image}" style="height:200px;"><br>
            <b>{title}</b><br>
            <i>{author}</i>
        </div>
        """
    
    html += "</div>"
    display(HTML(html))
```

```python
recommend('Harry Potter and the Prisoner of Azkaban (Book 3)')
```

### **Random Forest**

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import pandas as pd

# Prepare data
data = final_ratings.copy()
data['user_id'] = data['User-ID'].astype('category').cat.codes
data['book_id'] = data['ISBN'].astype('category').cat.codes
X = data[['user_id', 'book_id']]
y = data['Book-Rating']

# Train Random Forest
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)

# Map ISBNs to book_ids
book_codes = data[['ISBN', 'book_id']].drop_duplicates()
user_codes = data[['User-ID', 'user_id']].drop_duplicates().set_index('User-ID')['user_id'].to_dict()

# Create pseudo-ratings matrix (706 books × 810 users)
pseudo_ratings = np.zeros_like(pt)
for i, book_title in enumerate(pt.index):
    isbn_matches = books[books['Book-Title'] == book_title]['ISBN']
    if not isbn_matches.empty and isbn_matches.iloc[0] in book_codes['ISBN'].values:
        isbn = isbn_matches.iloc[0]
        book_id = book_codes[book_codes['ISBN'] == isbn]['book_id'].iloc[0]
        for j, user_id in enumerate(pt.columns):
            if user_id in user_codes:
                user_code = user_codes[user_id]
                input_df = pd.DataFrame([[user_code, book_id]], columns=['user_id', 'book_id'])
                pseudo_ratings[i, j] = model.predict(input_df)[0]

# Compute similarity between books
similarity_scores = cosine_similarity(pseudo_ratings)
```

```python
from IPython.display import HTML
import numpy as np

def recommend(book_name):
    index = np.where(pt.index == book_name)[0][0]
    similar_items = sorted(list(enumerate(similarity_scores[index])), key=lambda x: x[1], reverse=True)[1:5]
    
    selected_book = books[books['Book-Title'] == book_name].drop_duplicates('Book-Title')
    if selected_book.empty:
        print(f"Book '{book_name}' not found in books dataset.")
        return
    selected_title = selected_book['Book-Title'].values[0]
    selected_author = selected_book['Book-Author'].values[0]
    selected_image = selected_book['Image-URL-M'].values[0]
    
    html = f"""
    <h2>📚 Selected Book</h2>
    <div style='text-align: center; margin-bottom: 40px;'>
        <img src="{selected_image}" style="height:250px;"><br>
        <b>{selected_title}</b><br>
        <i>{selected_author}</i>
    </div>
    <h2>📚 Recommended Books</h2>
    <div style='display: flex; flex-wrap: wrap;'>
    """
    
    for i in similar_items:
        temp_df = books[books['Book-Title'] == pt.index[i[0]]].drop_duplicates('Book-Title')
        if temp_df.empty:
            continue
        title = temp_df['Book-Title'].values[0]
        author = temp_df['Book-Author'].values[0]
        image = temp_df['Image-URL-M'].values[0]
        
        html += f"""
        <div style="width: 22%; margin: 1%; text-align: center;">
            <img src="{image}" style="height:200px;"><br>
            <b>{title}</b><br>
            <i>{author}</i>
        </div>
        """
    
    html += "</div>"
    display(HTML(html))
```

```python
recommend('Harry Potter and the Prisoner of Azkaban (Book 3)')
```
